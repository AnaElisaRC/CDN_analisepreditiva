### RESPOSTA

Item (2A)

Ao mudar para o nível de cliente, a unidade de análise passa a ser o indivíduo (cliente_id), e não mais cada transação isolada. A variável-alvo agora indica se o cliente possui histórico de fraude (1 se ele teve pelo menos uma transação fraudulenta, 0 caso contrário). As variáveis explicativas precisam ser agregadas por cliente, como a soma total de compras e a quantidade de transações realizadas por cada um.



```
# RESPOSTA Item (2B)
# Seu código abaixo:

import pandas as pd

dados = {
    'transacao_id': [1,2,3,4,5,6,7,8],
    'cliente_id': [1,1,2,2,3,3,4,4],
    'valor_compra': [100,200,50,80,300,150,400,500],
    'tempo_cliente':[12,12,6,6,24,24,36,36],
    'valor_estornado': [0,200,0,0,300,0,0,500],
    'fraude': [0,1,0,0,1,0,0,1]
}
df_transacional = pd.DataFrame(dados)

df_agregado = df_transacional.groupby('cliente_id').agg({
    'valor_compra': 'sum',
    'transacao_id': 'count',
    'tempo_cliente': 'first',
    'fraude': 'max'
}).rename(columns={
    'valor_compra': 'soma_valor_compra',
    'transacao_id': 'quantidade_transacoes'
}).reset_index()

X = df_agregado.drop(columns=['cliente_id', 'fraude'])
y = df_agregado['fraude']


print("--- DataFrame Agregado (Nível Cliente) ---")
display(df_agregado)
print("\n--- Variáveis Explicativas (X) ---")
display(X)

print("\n--- Variável-Alvo (y) ---")
print(y.tolist())
```

## Questão 04 (1.0)

a) O que é codificação de variáveis categóricas e por que ela é necessária em modelos preditivos?

b) No dataset abaixo, aplique codificação One-Hot na variável tipo_pagamento.

    'cliente_id': [1,2,3,4],
    'tipo_pagamento': ['cartao', 'boleto', 'pix', 'cartao']


### RESPOSTA

Item (4A)

A codificação de variáveis categóricas é o processo de transformar valores categóricos em valores numéricos para que possam ser utilizados por algoritmos de aprendizado de máquina. Ela atua como um "tradutor" da linguagem humana para a linguagem de máquina. Ela é necessária porque a maioria dos modelos matemáticos não conseguem realizar cálculos e encontrar padrões se os dados estiverem em formato de texto.

```
# RESPOSTA Item (4B)
# Seu código abaixo:

import pandas as pd

dados_4b = {
    'cliente_id': [1,2,3,4],
    'tipo_pagamento': ['cartao', 'boleto', 'pix', 'cartao']
}
df_4b = pd.DataFrame(dados_4b)

df_4b_final = pd.get_dummies(df_4b, columns=['tipo_pagamento'], prefix='pagto')

print("--- Questão 4B (One-Hot Encoding) ---")
display(df_4b_final)
```

## Questão 05 (1.0)

a) O que é codificação ordinal e em que tipo de variável ela deve ser aplicada?

b) No dataset abaixo, aplique codificação ordinal na variável nivel_cliente, considerando a ordem: baixo < medio < alto.

    'cliente_id': [1,2,3,4],
    'nivel_cliente': ['baixo', 'alto', 'medio', 'baixo']

### RESPOSTA

Item (5A)

A codificação ordinal é simplesmente trocar categorias de texto por números em sequência (como 1, 2 e 3) para mostrar para o modelo que existe uma ordem de 'tamanho' ou 'importância' entre elas. Ela deve ser aplicada em variáveis categóricas ordinais, ou seja, aquelas que possuem uma ordem lógica ou hierarquia intrínseca (ex: tamanhos P/M/G, níveis de satisfação ou escolaridade). Se a ordem não importar (como 'cor do carro'), o ideal é usar One-Hot.

```
# RESPOSTA Item (5B)
# Seu código abaixo:

import pandas as pd

dados_5b = {
    'cliente_id': [1,2,3,4],
    'nivel_cliente': ['baixo', 'alto', 'medio', 'baixo']
}
df_5b = pd.DataFrame(dados_5b)

ordem = {
    'baixo': 1,
    'medio': 2,
    'alto': 3
}

df_5b['nivel_cliente_codificado'] = df_5b['nivel_cliente'].map(ordem)

print("--- Questão 5B (Codificação Ordinal) ---")
display(df_5b)
```

## Questão 06 (1.0)

a) O que é normalização e qual seu objetivo no contexto de modelos preditivos?

b) No dataset abaixo, aplique normalização (Min-Max) na variável renda, transformando os valores para o intervalo [0,1].

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]


### RESPOSTA

Item (6A)

A normalização é um tipo de escalonamento. Ela consiste em pegar valores distintos e reescalá-los em uma faixa numérica ou intervalo fixo, como entre 0 e 1. Sendo assim, o menor valor da variável passa a ser 0 e o maior 1, os demais são posicionados proporcionalemente entre este intervalor, isso reduz a complexidade de cálculo para o modelo. O objetivo principal é evitar que variáveis com valores absolutos muito grandes dominem algoritmos baseados em distância ou gradiente, colocando todos os atributos na mesma escala de grandeza e facilitando o aprendizado do modelo

```
# RESPOSTA Item (6B)
# Seu código abaixo:

import pandas as pd

dados_6b = {
    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]
}
df_6b = pd.DataFrame(dados_6b)

min_renda = df_6b['renda'].min()
max_renda = df_6b['renda'].max()

df_6b['renda_normalizada'] = (df_6b['renda'] - min_renda) / (max_renda - min_renda)

print("--- Questão 6B (Normalização Min-Max) ---")
display(df_6b)
```

## Questão 07 (1.0)

a) O que é padronização e qual a diferença em relação à normalização?

b) No dataset abaixo, aplique padronização (Z-score) na variável renda.

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]

### RESPOSTA

Item (7A)

 A padronização é uma forma de escalonamento que visa transformar as variáveis para uma mesma escala. Assim como na normalização, seu objetivo é garantir que nenhuma feature domine o modelo. A principal diferença em relação à normalização é que a padronização transforma os valores de uma variável para que tenham média 0 e desvio padrão 1. Dessa forma, cada valor passa a representar quantos desvios padrão ele está acima ou abaixo da média da variável.

```
# RESPOSTA Item (7B)
# Seu código abaixo:
import pandas as pd

dados_7b = {
    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]
}
df_7b = pd.DataFrame(dados_7b)

media_renda = df_7b['renda'].mean()
desvio_renda = df_7b['renda'].std(ddof=0)

df_7b['renda_padronizada'] = (df_7b['renda'] - media_renda) / desvio_renda

print("--- Questão 7B (Padronização Z-score) ---")
display(df_7b)
```

## Questão 10 (1.0)

a) O que é informação mútua e como ela pode ser utilizada na seleção de atributos?

b) Crie o DataFrame abaixo, calcule a informação mútua entre as variáveis explicativas e a variável-alvo e identifique qual variável apresenta menor relevância. Em seguida, remova essa variável, separe X e y e exiba o resultado.

    "temperatura": [15,18,21,24,27,30,33,36,39],
    "marketing": [100,120,140,160,180,200,220,240,260],
    "renda_regiao": [2000,2100,2200,2300,2400,2500,2600,2700,2800],
    "vendas_sorvete": [180,250,330,420,520,630,750,880,1020]

### RESPOSTA

Item (10A)

A informação mútua é uma medida que quantifica o quanto uma variável reduz a incerteza sobre outra, indicando o grau de dependência entre elas. Na seleção de atributos, ela é utilizada para identificar quais variáveis possuem maior relação com a variável-alvo, permitindo selecionar as mais informativas, inclusive quando existem relações não lineares.

```
# RESPOSTA Item (10B)
# Seu código abaixo:

import pandas as pd
from sklearn.feature_selection import mutual_info_regression

dados_10b = {
   "temperatura": [15,18,21,24,27,30,33,36,39],
"marketing": [100,120,140,160,180,200,220,240,260],
"renda_regiao": [2000,2100,2200,2300,2400,2500,2600,2700,2800],
"vendas_sorvete": [180,250,330,420,520,630,750,880,1020]
}
df_10b = pd.DataFrame(dados_10b)

X_total = df_10b.drop(columns=['vendas_sorvete'])
y_alvo = df_10b['vendas_sorvete']


mi = mutual_info_regression(X_total, y_alvo, random_state=42)

mi_df = pd.DataFrame({'Atributo': X_total.columns, 'Informação Mútua': mi})
print("--- Informação Mútua (MI) ---")
display(mi_df.sort_values(by='Informação Mútua', ascending=False))

X_final = X_total.drop(columns=['renda_regiao'])

print("\nVariáveis explicativas (X) após remover a menor relevância (renda_regiao):")
display(X_final)

print("\nVariável-alvo (y):")
print(y_alvo.tolist())
```
