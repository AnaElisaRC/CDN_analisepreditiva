Construa um novo DataFrame com:

### RESPOSTA

Item (2A)
Ao reformular o problema para o nível de cliente, a estrutura do seu projeto preditivo muda da seguinte forma:
###Unidade de Análise:

A entidade fundamental sobre a qual a previsão será feita muda da "transação" para o "cliente"

Isso significa que a granularidade do banco de dados é alterada para que cada linha represente estritamente um cliente único, agregando todo o seu histórico

###Variável-Alvo:
O alvo deixa de ser focado num evento isolado (prever se a transação 123 é fraude) e passa a focar-se no comportamento do indivíduo (ex: prever se o cliente cometerá alguma fraude ou a quantidade de fraudes que tentará realizar)

Variáveis Explicativas: Como a unidade de análise subiu um nível hierárquico, as variáveis pontuais de eventos (como o valor de uma única compra) deixam de fazer sentido isoladamente
É necessário aplicar técnicas de agregação (Engenharia de Atributos) para resumir a informação de vários registos num único atributo para aquele cliente, utilizando operações matemáticas como soma, contagem, média ou valor máximo

```python
# RESPOSTA Item (2B)
# 1. Criação do dicionário com os dados originais
dados = {
    'transacao_id': [1,2,3,4,5,6,7,8],
    'cliente_id':   [1,1,2,2,3,3,4,4],
    'valor_compra': [100,200,50,80,300,150,400,500],
    'tempo_cliente':[12,12,6,6,24,24,36,36],
    'valor_estornado': [0,200,0,0,300,0,0,500],
    'fraude':       [0,1,0,0,1,0,0,1]
}

# Criação do DataFrame original
df = pd.DataFrame(dados)

# 2. Agregação dos dados por cliente (Engenharia de Atributos)
# Agrupamos por 'cliente_id' e definimos como cada coluna deve ser calculada
df_clientes = df.groupby('cliente_id').agg(
    soma_valor_compra=('valor_compra', 'sum'),       # Soma de valor_compra
    qtd_transacoes=('transacao_id', 'count'),        # Quantidade de transações
    tempo_cliente=('tempo_cliente', 'first'),        # Manter o primeiro valor do tempo_cliente
    alvo_fraude=('fraude', 'max')                    # 1 se teve ao menos uma fraude, 0 caso contrário
).reset_index()

# 3. Definição da variável-alvo (y)
y = df_clientes['alvo_fraude']

# 4. Separação das variáveis explicativas (X)
# Removemos o identificador do cliente (irrelevante) e o próprio alvo para criar o X
X = df_clientes.drop(columns=['cliente_id', 'alvo_fraude'])

# Exibindo os resultados finais
print("=== DataFrame Agregado (Visão Cliente) ===")
print(df_clientes)

print("\n=== Variáveis Explicativas (X) ===")
print(X)

print("\n=== Variável-Alvo (y) ===")
print(y)
```

## Questão 04 (1.0)

a) O que é codificação de variáveis categóricas e por que ela é necessária em modelos preditivos?

b) No dataset abaixo, aplique codificação One-Hot na variável tipo_pagamento.

    'cliente_id': [1,2,3,4],
    'tipo_pagamento': ['cartao', 'boleto', 'pix', 'cartao']


### RESPOSTA

Item (4A)
A codificação de variáveis categóricas é o processo de transformar valores categóricos (como rótulos, nomes ou classes em formato de texto) em representações estritamente numéricas

Ela é absolutamente necessária porque a grande maioria dos algoritmos de aprendizado de máquina e modelos estatísticos não consegue interpretar palavras ou textos soltos

Como estes modelos operam através de equações e cálculos matemáticos, exigem que todas as entradas de dados (variáveis explicativas) sejam matrizes ou valores numéricos para conseguirem processar as informações e realizar o treinamento de forma correta

```python
# RESPOSTA Item (4B)
# Seu código abaixo:
import pandas as pd

# 1. Criação do dataset original
dados = {
    'cliente_id': [1,2,3,4],
    'tipo_pagamento': ['cartao', 'boleto', 'pix', 'cartao']
}
df = pd.DataFrame(dados)

# 2. Aplicação da codificação One-Hot
# O método get_dummies converte a variável categórica em colunas binárias
# O parâmetro dtype=int garante que os valores sejam 0 e 1 (em vez de True/False)
df_codificado = pd.get_dummies(df, columns=['tipo_pagamento'], dtype=int)

# Exibindo o resultado final
print(df_codificado)
```

## Questão 06 (1.0)

b) No dataset abaixo, aplique normalização (Min-Max) na variável renda, transformando os valores para o intervalo [0,1].

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]


### RESPOSTA

Item (6A)
Normalização é basicamente você colocar todos os números na mesma escala, geralmente entre 0 e 1, sem perder a diferença real entre eles.

Pensa num exemplo: você tem um modelo que analisa casas com dois dados — número de quartos (de 1 a 5) e área (de 50 a 500 m²). Se você jogar isso direto no modelo, ele pode achar que a área é muito mais importante só porque os números são maiores.

A normalização resolve isso colocando tudo na mesma “régua”, aí o modelo consegue comparar de forma mais justa e aprender melhor o que realmente importa.

```python
# RESPOSTA Item (6B)
# Seu código abaixo:
import pandas as pd

df = pd.DataFrame({'cliente_id': [1,2,3,4], 'renda': [2000, 5000, 3000, 4000]})

# Aplicando a fórmula Min-Max
df['renda_norm'] = (df['renda'] - df['renda'].min()) / (df['renda'].max() - df['renda'].min())

print(df)
```

## Questão 07 (1.0)

a) O que é padronização e qual a diferença em relação à normalização?

b) No dataset abaixo, aplique padronização (Z-score) na variável renda.

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]

### RESPOSTA

Item (7A)

A padronização (também conhecida como Z-score) é um tipo de transformação matemática que padroniza a escala das variáveis numéricas para que estas passem a ter média igual a 0 e desvio padrão igual a 1
Na prática, cada valor transformado passa a representar exatamente quantos desvios-padrão ele se encontra acima ou abaixo da média original da variável

A principal diferença em relação à normalização (Min-Max) reside no intervalo numérico resultante: a normalização comprime e reescala os valores obrigatoriamente para dentro de um intervalo fixo, geralmente entre 0 e 1
Por outro lado, a padronização não produz um intervalo fixo predeterminado de valores
Geralmente, a padronização é a técnica mais utilizada em algoritmos de Machine Learning sensíveis à escala, enquanto a normalização é preferida quando é estritamente necessário limitar os valores de entrada para evitar saturação (como em funções sigmoid)

Nenhuma das duas técnicas altera a forma original da distribuição dos dados

```python
# RESPOSTA Item (7B)
# Seu código abaixo:
import pandas as pd

# 1. Criação do DataFrame com o dataset fornecido
dados = {
    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]
}
df = pd.DataFrame(dados)

# 2. Cálculo da média e do desvio padrão da coluna 'renda'
media = df['renda'].mean()          # A média será 3500
desvio_padrao = df['renda'].std()   # O desvio padrão amostral será aprox. 1290.99

# 3. Aplicação da fórmula de padronização (Z-score)
df['renda_padronizada'] = (df['renda'] - media) / desvio_padrao

# Exibindo o resultado final
print(df)
```

Item (8A)
### RESPOSTA Item (8A)
É o processo de escolher um subconjunto a partir das variáveis originais que sejam mais relevantes para a construção do modelo preditivo. Tem como principal objetivo simplificar o modelo e focar no que realmente impacta a variável alvo para melhorar acurácia, além de reduzir alguns problemas, como: overfiting, tempo de treinamento e redundância de variáveis.

## Questão 09 (1.0)

a) Como a correlação pode ser utilizada no processo de seleção de atributos?

b) Crie o DataFrame abaixo, calcule a matriz de correlação e, com base nela, identifique qual variável pode ser removida por redundância. Em seguida, remova essa variável, separe X e y e exiba o resultado.

    "idade": [22, 25, 28, 35, 45, 52, 30, 40],
    "renda": [2500, 3200, 4100, 5200, 7000, 8200, 4500, 6100],
    "anos_estudo": [12, 14, 16, 18, 16, 14, 15, 17],
    "tempo_empresa": [1, 2, 3, 6, 10, 15, 4, 8],
    "salario": [3000, 3500, 4500, 6000, 7200, 8000, 5000, 6500]

### RESPOSTA

Item (9A)
Correlação é uma das formas mais simples e fortes de entender se uma variável realmente ajuda o modelo ou só tá ali ocupando espaço. Basicamente, ela mostra o quanto duas coisas estão relacionadas.

Na prática, você usa isso pra “limpar” o dataset com dois focos principais:

Primeiro: ver o que realmente importa.
Você olha quais variáveis têm uma relação forte com aquilo que você quer prever. Se a correlação for próxima de zero, é um sinal de que aquela variável não ajuda em nada — é só ruído. Então a ideia é manter o que realmente tem impacto no resultado.

Segundo: evitar redundância.
Aqui é onde muita gente se perde. Se duas variáveis são muito parecidas (tipo renda mensal e salário anual), elas praticamente dizem a mesma coisa. Isso não ajuda o modelo — pelo contrário, deixa ele mais pesado, mais lento e até pode atrapalhar o resultado.

```python
# RESPOSTA Item (9B)
# Seu código abaixo:
import pandas as pd

# 1. Criar o DataFrame
data = {
    "idade": [22, 25, 28, 35, 45, 52, 30, 40],
    "renda": [2500, 3200, 4100, 5200, 7000, 8200, 4500, 6100],
    "anos_estudo": [12, 14, 16, 18, 16, 14, 15, 17],
    "tempo_empresa": [1, 2, 3, 6, 10, 15, 4, 8],
    "salario": [3000, 3500, 4500, 6000, 7200, 8000, 5000, 6500]
}

df = pd.DataFrame(data)

# 2. Calcular a matriz de correlação
matriz_corr = df.corr()
print("Matriz de Correlação:")
print(matriz_corr)
print("-" * 30)

# 3. Identificação da redundância:
# Observando a matriz, 'renda' e 'salario' possuem uma correlação extremamente alta (0.99).
# Elas representam praticamente a mesma informação financeira.
variavel_redundante = 'renda'

# 4. Remover a variável redundante
df_limpo = df.drop(columns=[variavel_redundante])

# 5. Separar X (preditores) e y (alvo)
# Vamos assumir que o objetivo é prever o 'salario' (y) com base nos outros dados (X)
X = df_limpo.drop(columns=['salario'])
y = df_limpo['salario']

print("Variáveis Preditoras (X):")
print(X)
print("\nVariável Alvo (y):")
print(y)
```

## Questão 10 (1.0)

a) O que é informação mútua e como ela pode ser utilizada na seleção de atributos?

b) Crie o DataFrame abaixo, calcule a informação mútua entre as variáveis explicativas e a variável-alvo e identifique qual variável apresenta menor relevância. Em seguida, remova essa variável, separe X e y e exiba o resultado.

    "temperatura": [15,18,21,24,27,30,33,36,39],
    "marketing": [100,120,140,160,180,200,220,240,260],
    "renda_regiao": [2000,2100,2200,2300,2400,2500,2600,2700,2800],
    "vendas_sorvete": [180,250,330,420,520,630,750,880,1020]

### RESPOSTA

Item (10A)
A informação mútua é uma métrica estatística que mede o quanto o conhecimento do valor de uma variável reduz a incerteza sobre o valor de outra
Ela é definida com base nas probabilidades conjuntas e marginais das variáveis, sendo que um valor igual a zero indica que as variáveis são completamente independentes, e valores maiores indicam uma forte dependência entre elas

Na etapa de seleção de atributos (Engenharia de Atributos), a informação mútua é utilizada para ranquear e avaliar o quanto cada variável explicativa contribui para explicar a variável-alvo
O seu principal diferencial e vantagem em relação à correlação tradicional é que a informação mútua não se limita a identificar relações lineares, sendo capaz de detetar dependências muito mais complexas, não lineares e não monotônicas entre os dados

Dessa forma, ela orienta a seleção: variáveis com alta pontuação de informação mútua em relação ao alvo devem ser priorizadas e mantidas no modelo, enquanto aquelas com pontuações próximas de zero (que indicam ausência de relação e funcionam como ruído) são descartadas para reduzir a dimensionalidade


```python
import pandas as pd
from sklearn.feature_selection import mutual_info_regression

# 1. Criação do DataFrame
data = {
    "temperatura":    [15, 18, 21, 24, 27, 30, 33, 36, 39],
    "marketing":      [100, 120, 140, 160, 180, 200, 220, 240, 260],
    "renda_regiao":   [2000, 2100, 2200, 2300, 2400, 2500, 2600, 2700, 2800],
    "vendas_sorvete": [180, 250, 330, 420, 520, 630, 750, 880, 1020]
}

df = pd.DataFrame(data)

# 2. Separando variáveis explicativas e alvo para o cálculo
X_temp = df.drop(columns=['vendas_sorvete'])
y_temp = df['vendas_sorvete']

# 3. Calculando a Informação Mútua
# O discrete_features=False indica que as variáveis são contínuas
mi_scores = mutual_info_regression(X_temp, y_temp, discrete_features=False, random_state=42)
mi_series = pd.Series(mi_scores, index=X_temp.columns)

print("=== Scores de Informação Mútua ===")
print(mi_series.sort_values(ascending=False))

# 4. Identificando a variável de menor relevância
# (Neste dataset específico, as variáveis são perfeitamente lineares entre si,
# mas o MI identificará qual tem a menor dependência estatística detectada)
variavel_menos_relevante = mi_series.idxmin()
print(f"\nVariável removida: {variavel_menos_relevante}")

# 5. Removendo a variável, separando X e y
X = X_temp.drop(columns=[variavel_menos_relevante])
y = y_temp

# 6. Exibindo o resultado final
print("\n=== DataFrame Final (X) ===")
print(X)
print("\n=== Alvo (y) ===")
print(y.values)
```
