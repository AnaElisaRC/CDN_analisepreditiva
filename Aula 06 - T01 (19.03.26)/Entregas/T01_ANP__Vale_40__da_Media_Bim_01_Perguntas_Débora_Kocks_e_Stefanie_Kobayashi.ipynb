Item (2A)
A unidade de análise que antes era por transação passa a ser por cliente. E a variável-alvo agora é fraude do cliente, antes era fraude por transação. As variáveis explicativas mudam, porque agora estamos olhando valores agregados e não individuais

```
# RESPOSTA Item (2B)
# Seu código abaixo:

import pandas as pd

# Criando o DataFrame
df2 = pd.DataFrame({
    'transacao_id': [1,2,3,4,5,6,7,8],
    'cliente_id': [1,1,2,2,3,3,4,4],
    'valor_compra': [100,200,50,80,300,150,400,500],
    'tempo_cliente': [12,12,6,6,24,24,36,36],
    'valor_estornado': [0,200,0,0,300,0,0,500],
    'fraude': [0,1,0,0,1,0,0,1]
})

df21 = df2.groupby('cliente_id').agg({
    'valor_compra': 'sum',
    'transacao_id': 'count',
    'tempo_cliente': 'first',
    'fraude': 'max'
}).reset_index()

# Separando X e y
X = df21.drop(columns=['cliente_id', 'fraude'])
y = df21['fraude']

print(X)
print(y)
```

### RESPOSTA

Item (6A)
É uma técnica de processamento de dados que serve para colocar valores numéricos na mesma escala, facilitando a comparação e o uso em modelos de machine learning

```
# RESPOSTA Item (6B)
# Seu código abaixo:

df6 = pd.DataFrame({
    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]
})

# Normalização Min-Max.
df6["renda_normalizada"] = (df6["renda"] - df6["renda"].min()) / (df6["renda"].max() - df6["renda"].min())

df6[["renda","renda_normalizada"]]
```

## Questão 07 (1.0)

a) O que é padronização e qual a diferença em relação à normalização?

b) No dataset abaixo, aplique padronização (Z-score) na variável renda.

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]

### RESPOSTA

Item (7A)
A padronização transforma os valores de uma variável para que passem a ter média 0 e desvio padrão 1. A padronização também não altera a forma da distribuição dos dados e reescala os valores da variável em
relação à média e ao desvio padrão.  

Diferente da normalização, ela não limita os dados a um intervalo fixo e é menos sensível a outliers

```
import pandas as pd
from sklearn.preprocessing import StandardScaler

df7 = pd.DataFrame({
    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]
})

# Aplicar padronização (StandardScaler).
scaler_std = StandardScaler()
df7['renda_padronizada'] = scaler_std.fit_transform(df7[['renda']])

print(df7[['renda', 'renda_padronizada']])
```

### RESPOSTA

Item (9A)
A correlação pode ser utilizada para identificar variáveis altamente correlacionadas entre si ou pouco relacionadas com a variável-alvo

```
# RESPOSTA Item (9B)
# Seu código abaixo:


df9 = pd.DataFrame({
    "idade": [22, 25, 28, 35, 45, 52, 30, 40],
    "renda": [2500, 3200, 4100, 5200, 7000, 8200, 4500, 6100],
    "anos_estudo": [12, 14, 16, 18, 16, 14, 15, 17],
    "tempo_empresa": [1, 2, 3, 6, 10, 15, 4, 8],
    "salario": [3000, 3500, 4500, 6000, 7200, 8000, 5000, 6500]
})

df9.corr(numeric_only=True)["salario"].sort_values(ascending=False)



# Removendo variável redundante (renda, altamente correlacionada com salario)
df91 = df9.drop(columns=['renda'])

# Separando X e y (considerando salario como alvo)
X = df91.drop(columns=['salario'])
y = df91['salario']

print("\nX:\n", X)
print("-" * 30)
print("\ny:\n", y)
```

## Questão 10 (1.0)

a) O que é informação mútua e como ela pode ser utilizada na seleção de atributos?

b) Crie o DataFrame abaixo, calcule a informação mútua entre as variáveis explicativas e a variável-alvo e identifique qual variável apresenta menor relevância. Em seguida, remova essa variável, separe X e y e exiba o resultado.

    "temperatura": [15,18,21,24,27,30,33,36,39],
    "marketing": [100,120,140,160,180,200,220,240,260],
    "renda_regiao": [2000,2100,2200,2300,2400,2500,2600,2700,2800],
    "vendas_sorvete": [180,250,330,420,520,630,750,880,1020]

### RESPOSTA

Item (10A)
A informação mútua mede o quanto uma variável ajuda a explicar a outra. Na seleção de atributos, indica quais variáveis são mais úteis para prever a variável-alvo

```
# RESPOSTA Item (10B)
# Seu código abaixo:

df10 = pd.DataFrame({
    "temperatura": [15,18,21,24,27,30,33,36,39],
    "marketing": [100,120,140,160,180,200,220,240,260],
    "renda_regiao": [2000,2100,2200,2300,2400,2500,2600,2700,2800],
    "vendas_sorvete": [180,250,330,420,520,630,750,880,1020]
})

from sklearn.feature_selection import mutual_info_regression
import pandas as pd

X_temp = df10.drop(columns=['vendas_sorvete'])
y_temp = df10['vendas_sorvete']

mi_scores = mutual_info_regression(X_temp, y_temp, random_state=42)
mi_series = pd.Series(mi_scores, index=X_temp.columns).sort_values(ascending=True)

print("Informação Mútua de cada variável com 'vendas_sorvete':")
print(mi_series)

# Identificar a variável com menor relevância
least_relevant_variable = mi_series.index[0]
print(f"\nA variável com menor relevância é: {least_relevant_variable}")

# Remover a variável identificada
X = X_temp.drop(columns=[least_relevant_variable])
y = y_temp

print("\nX após remoção da variável menos relevante:")
print(X)
print("\ny (variável-alvo):")
print(y)
```
