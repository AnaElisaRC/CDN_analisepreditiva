Construa um novo DataFrame com:

soma de valor_compra

### RESPOSTA

2a) Ao reformular o problema para o nivel de cliente, todas as variáveis devem ser reeinterpretadas. Portanto, considere-se:


| Elemento | Nível transação | Nível cliente |
| :-- | :-- | :-- |
| Unidade de análise | Uma transação | Um cliente |
| Variável-alvo | `fraude` da transação | `fraude_cliente` (1 se teve ao menos 1 fraude) |
| Variáveis explicativas | `valor_compra`, `tempo_cliente` por linha | `total_valor_compra`, `qtd_transacoes`, `tempo_cliente` agregados |


Em síntese, ao nível do cliente, a unidade de análise é "um cliente", a variável - alvo é "fraude- cliente", e as variáveis explicativas tornam-se "total_valor_compra", "qtd_transacoes" e "tempo_cliente"


```
# RESPOSTA Item (2B)

df_agg = df.groupby('cliente_id').agg(
    total_valor_compra=('valor_compra', 'sum'),
    qtd_transacoes=('transacao_id', 'count'),
    tempo_cliente=('tempo_cliente', 'first'),
    fraude_cliente=('fraude', 'max')   # 1 se teve ao menos uma fraude
).reset_index()

X = df_agg.drop(columns=['cliente_id', 'fraude_cliente'])
y = df_agg['fraude_cliente']

Xy = pd.concat([X, y], axis=1)
print(Xy)
```

É a criação de novas variáveis a partir das variáveis existentes com o objetivo de capturar relações, padrões ou estruturas que não estão explícitos nos dados originais. Envolve gerar atributos derivados por meio de combinações, agregações ou transformações simples das colunas.

## Questão 05 (1.0)

a) O que é codificação ordinal e em que tipo de variável ela deve ser aplicada?

b) No dataset abaixo, aplique codificação ordinal na variável nivel_cliente, considerando a ordem: baixo < medio < alto.

    'cliente_id': [1,2,3,4],
    'nivel_cliente': ['baixo', 'alto', 'medio', 'baixo']

### RESPOSTA

Item (5A)  
A codificação ordinal consiste em substituir categorias por valores numéricos inteiros que representem uma hierarquia previamente existente. Há imposição explícita de ordem entre as categorias. Exemplo: júnior, pleno e sênior (1, 2 e 3).

```
# RESPOSTA Item (5B)

import pandas as pd

# Criando o DataFrame
dados = {
    'cliente_id': [1, 2, 3, 4],
    'nivel_cliente': ['baixo', 'alto', 'medio', 'baixo']
}
df = pd.DataFrame(dados)

# Definindo a ordem (Dicionário de Mapeamento)
ordem_niveis = {
    'baixo': 1,
    'medio': 2,
    'alto': 3
}

# Aplicando a codificação ordinal
df['nivel_cliente_cod'] = df['nivel_cliente'].map(ordem_niveis)

print(df)
```

## Questão 06 (1.0)

a) O que é normalização e qual seu objetivo no contexto de modelos preditivos?

b) No dataset abaixo, aplique normalização (Min-Max) na variável renda, transformando os valores para o intervalo [0,1].

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]


### RESPOSTA

Item (6A)  
A normalização (Min–Max) reescala os valores de uma variável numérica para um intervalo fixo, geralmente [0,1]. O menor valor da variável passa a ser 0 e o maior valor passa a ser 1, enquanto os demais valores são posicionados proporcionalmente entre esses limites. Esse procedimento é útil quando diferentes variáveis possuem escalas muito distintas, pois coloca todas na mesma faixa numérica sem alterar a ordem relativa dos valores. Um exemplo prático é a comparação de notas em escalas diferentes: um aluno que tira 8 em uma prova de 10 e outro que tira 72 em uma de 80. Ao normalizarmos ambos para o intervalo [0,1], obtemos 0,8 e 0,9, respectivamente, permitindo que o modelo identifique qual desempenho foi proporcionalmente superior sem ser enganado pela magnitude dos números.

```
# RESPOSTA Item (6B)

import pandas as pd

# Criando o DataFrame
dados = {
    'cliente_id': [1, 2, 3, 4],
    'renda': [2000, 5000, 3000, 4000]
}
df = pd.DataFrame(dados)

# Normalização Min-Max.
df["renda_normalizada"] = (df["renda"] - df["renda"].min()) / (df["renda"].max() - df["renda"].min())

df[["renda","renda_normalizada"]]
```

## Questão 07 (1.0)

a) O que é padronização e qual a diferença em relação à normalização?

b) No dataset abaixo, aplique padronização (Z-score) na variável renda.

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]

### RESPOSTA

7a) A Padronização (Z-Score) transforma os valores para que a variável passe a ter *média 0 e desvio padrão 1*. *Cada valor representa quantos desvios-padrão está acima ou abaixo da média*. A Normalização (Min - Max), por sua vez, reescala os valores para um intervalo fixo. *O menor valor vira 0 e o maior vira 1;* os demais são posicionados proporcionalmente.*

Ambas são tipos de Escalonamento ( colocar as variáveis em escalas comparáveis), mas enquanto a Normalização coloca os valores em um intervalo fixo (normalmente de 0 a 1), a Padronização centraliza os dados na média e ajusta a dispersão.

```
# RESPOSTA Item (7B)

from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({'cliente_id': [1,2,3,4], 'renda': [2000, 5000, 3000, 4000]})

scaler = StandardScaler()
df['renda_zscore'] = scaler.fit_transform(df[['renda']])

display(df)

```

## Questão 08 (1.0)

a) O que é seleção de atributos e qual seu objetivo em análise preditiva?

b) No dataset abaixo, identifique e remova as variáveis que não devem ser utilizadas no modelo e, em seguida, separe X e y.

    'cliente_id': [1,2,3,4],
    'idade': [25, 40, 35, 50],
    'renda_mensal': [2000, 5000, 3000, 4000],
    'renda_anual': [24000, 60000, 36000, 48000],
    'inadimplente': [0,1,0,1]

### RESPOSTA

Item (8A)  
A seleção de atributos consiste em identificar e manter apenas as variáveis originais mais relevantes de um dataset, removendo aquelas que são redundantes ou irrelevantes. Diferente da construção de atributos, ela não transforma os dados, mas seleciona um subconjunto das colunas existentes. O objetivo principal é simplificar o modelo, reduzir o tempo de treinamento, evitar o overfitting e melhorar a interpretabilidade dos resultados, garantindo que o algoritmo foque apenas no 'sinal' que realmente impacta a variável alvo.

```
# RESPOSTA Item (8B)

import pandas as pd

# Criando o DataFrame (seu código base)
dados = {
    'cliente_id': [1,2,3,4],
    'idade': [25, 40, 35, 50],
    'renda_mensal': [2000, 5000, 3000, 4000],
    'renda_anual': [24000, 60000, 36000, 48000],
    'inadimplente': [0,1,0,1]
}
df = pd.DataFrame(dados)

# Removendo as variáveis irrelevantes ou redundantes
df_limpo = df.drop(columns=['cliente_id', 'renda_anual'])

# Separando X (atividades/features) e y (alvo/target)
X = df_limpo.drop(columns=['inadimplente'])
y = df_limpo['inadimplente']

# Visualizando os resultados
print("Variáveis preditoras (X):")
print(X)
print("\nVariável alvo (y):")
print(y)
```

## Questão 09 (1.0)

a) Como a correlação pode ser utilizada no processo de seleção de atributos?

b) Crie o DataFrame abaixo, calcule a matriz de correlação e, com base nela, identifique qual variável pode ser removida por redundância. Em seguida, remova essa variável, separe X e y e exiba o resultado.

    "idade": [22, 25, 28, 35, 45, 52, 30, 40],
    "renda": [2500, 3200, 4100, 5200, 7000, 8200, 4500, 6100],
    "anos_estudo": [12, 14, 16, 18, 16, 14, 15, 17],
    "tempo_empresa": [1, 2, 3, 6, 10, 15, 4, 8],
    "salario": [3000, 3500, 4500, 6000, 7200, 8000, 5000, 6500]

### RESPOSTA

Item (9A)  
A correlação mede o grau de relação linear entre duas variáveis. É uma ferramenta essencial na seleção de atributos por dois motivos:

*   Filtro de Relevância: Ajuda a descartar variáveis que possuem correlação quase nula com a variável-alvo, pois elas não agregam poder preditivo ao modelo.

*   Redução de Redundância: Identifica variáveis preditoras que são altamente correlacionadas entre si. Nesses casos, as variáveis 'explicam a mesma coisa', e manter ambas pode causar instabilidade no modelo e aumentar o custo computacional desnecessariamente.

Ao remover uma das variáveis redundantes, simplificamos o modelo (tornando-o mais interpretável) sem perder a qualidade da informação original.

```
# RESPOSTA Item (9B)

import pandas as pd

# Criando o DataFrame
dados = {
    'idade': [22, 25, 28, 35, 45, 52, 30, 40],
    'renda': [2500, 3200, 4100, 5200, 7000, 8200, 4500, 6100],
    'anos_estudo': [12, 14, 16, 18, 16, 14, 15, 17],
    'tempo_empresa': [1, 2, 3, 6, 10, 15, 4, 8],
    'salario': [3000, 3500, 4500, 6000, 7200, 8000, 5000, 6500]
}
df = pd.DataFrame(dados)

# Calcular a matriz de correlação
matriz_corr = df.corr()
print("Matriz de Correlação:")
print(matriz_corr)

# Removendo a variável redundante
df_reduzido = df.drop(columns=['renda'])

# Separando X e y
# (Considerando 'salario' como a variável que queremos prever)
X = df_reduzido.drop(columns=['salario'])
y = df_reduzido['salario']

print("\nFeatures (X):")
print(X.head())
print("\nAlvo (y):")
print(y.head())
```

## Questão 10 (1.0)

a) O que é informação mútua e como ela pode ser utilizada na seleção de atributos?

b) Crie o DataFrame abaixo, calcule a informação mútua entre as variáveis explicativas e a variável-alvo e identifique qual variável apresenta menor relevância. Em seguida, remova essa variável, separe X e y e exiba o resultado.

    "temperatura": [15,18,21,24,27,30,33,36,39],
    "marketing": [100,120,140,160,180,200,220,240,260],
    "renda_regiao": [2000,2100,2200,2300,2400,2500,2600,2700,2800],
    "vendas_sorvete": [180,250,330,420,520,630,750,880,1020]

### RESPOSTA

10a) Informação mútua mede *o quanto o conhecimento de uma variável reduz a incerteza sobre a variável-alvo*. Diferente da correlação, ela também detecta relações *não-lineares*. Na seleção de atributos, calcula-se a informação mútua de cada feature com o alvo: variáveis com valor baixo contribuem pouco para a previsão e podem ser removidas.

```
# RESPOSTA Item (10B)

from sklearn.feature_selection import mutual_info_regression

df = pd.DataFrame({
    'temperatura':   [15,18,21,24,27,30,33,36,39],
    'marketing':     [100,120,140,160,180,200,220,240,260],
    'renda_regiao':  [2000,2100,2200,2300,2400,2500,2600,2700,2800],
    'vendas_sorvete':[180,250,330,420,520,630,750,880,1020]
})

X_all = df.drop(columns=['vendas_sorvete'])
y = df['vendas_sorvete']

mi = mutual_info_regression(X_all, y, random_state=42)
mi_series = pd.Series(mi, index=X_all.columns).sort_values()
print(mi_series)

# Remove a variável com menor informação mútua
X = X_all.drop(columns=[mi_series.idxmin()])
```
