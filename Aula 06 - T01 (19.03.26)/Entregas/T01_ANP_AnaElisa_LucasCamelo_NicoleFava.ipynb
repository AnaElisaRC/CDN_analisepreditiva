A unidade de análise é a transação, pois o enunciado define que cada linha do conjunto de dados representa um evento de compra individual. Essa unidade determina a granularidade e como os dados estão organizados.A variável-alvo é a coluna 'fraude'. Já a coluna valor_estornado deve ser excluída por configurar vazamento de informação (leakage), pois é um dado registrado apenas após a ocorrência do evento, tornando o modelo inviável para previsões em tempo real.

### RESPOSTA

Item (2A)

Ao reformular o problema para o nível de cliente, a estrutura do projeto de análise de dados sofre as seguintes alterações:Unidade de Análise: Deixa de ser o evento individual (transação) e passa a ser o cliente (entidade). Isso significa que cada linha da base de dados representará um cliente único, consolidando seu histórico.Variável-alvo: O alvo é redefinido para refletir o comportamento do cliente no tempo. Em vez de "esta transação é fraude?", a pergunta passa a ser "este cliente cometeu ou sofreu fraude em seu histórico?".Variáveis Explicativas: A forma de construção muda radicalmente, exigindo agregações. Atributos diretos da transação (como o valor de uma única compra) precisam ser transformados em métricas resumidas, como a média de gastos ou o total de transações do cliente.

```
# RESPOSTA Item (2B)
# Seu código abaixo:

import pandas as pd

# 1. Criação do DataFrame original
data = {
    'transacao_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'cliente_id': [1, 1, 2, 2, 3, 3, 4, 4],
    'valor_compra': [100, 200, 50, 80, 300, 150, 400, 500],
    'tempo_cliente': [12, 12, 6, 6, 24, 24, 36, 36],
    'valor_estornado': [0, 200, 0, 0, 300, 0, 0, 500],
    'fraude': [0, 1, 0, 0, 1, 0, 0, 1]
}
df = pd.DataFrame(data)

# 2. Agregação por cliente_id
df_cliente = df.groupby('cliente_id').agg({
    'valor_compra': 'sum',          # Soma de valor_compra
    'transacao_id': 'count',        # Quantidade de transações
    'tempo_cliente': 'first',       # Primeiro valor por cliente
    'fraude': 'max'                 # 1 se teve ao menos uma fraude, 0 caso contrário
}).rename(columns={'transacao_id': 'qtd_transacoes', 'valor_compra': 'total_gasto'})

# 3. Separação de X e y
# Nota: 'valor_estornado' foi excluído conforme solicitado para evitar leakage [cite: 95]
X = df_cliente[['total_gasto', 'qtd_transacoes', 'tempo_cliente']]
y = df_cliente['fraude']

print("DataFrame Agregado (X):")
print(X)
print("\nVariável-alvo (y):")
print(y)
```

## Questão 03 (1.0)

a) O que é construção de atributos em análise preditiva?

b) No dataset abaixo, crie uma nova variável chamada valor_medio_por_compra dividindo o total gasto pela quantidade de compras.

    'cliente_id': [1,2,3,4],
    'total_gasto': [1000, 2000, 1500, 3000],
    'qtd_compras': [5, 10, 3, 6]


### RESPOSTA

Item (3A)

A construção de atributos (feature engineering) é o processo de criar novas variáveis (atributos) a partir dos dados brutos existentes para melhorar o desempenho e a precisão dos modelos preditivos.

Enquanto a seleção de atributos remove o que é irrelevante, a construção busca extrair conhecimento oculto. Muitas vezes, a relação entre duas variáveis isoladas não explica bem o fenômeno, mas uma operação matemática entre elas revela um padrão crucial.

Seus principais objetivos são:
  * Capturar relações não lineares: Transformar dados brutos em indicadores mais robustos (ex: converter "data de nascimento" em "idade").

  * Isolar sinais importantes: Criar métricas de eficiência ou proporção que o modelo teria dificuldade em calcular sozinho.

  * Simplificar o aprendizado: Fornecer ao algoritmo uma "ponte" direta para o que realmente importa para a previsão.

```
# RESPOSTA Item (3B)
# Seu código abaixo:
import pandas as pd

# 1. Criando o dataset
data = {
    'cliente_id': [1, 2, 3, 4],
    'total_gasto': [1000, 2000, 1500, 3000],
    'qtd_compras': [5, 10, 3, 6]
}

df = pd.DataFrame(data)

# 2. Construindo o novo atributo: valor_medio_por_compra
# Esta é uma métrica de 'ticket médio'
df['valor_medio_por_compra'] = df['total_gasto'] / df['qtd_compras']

print(df)
```

## Questão 04 (1.0)

a) O que é codificação de variáveis categóricas e por que ela é necessária em modelos preditivos?

b) No dataset abaixo, aplique codificação One-Hot na variável tipo_pagamento.

    'cliente_id': [1,2,3,4],
    'tipo_pagamento': ['cartao', 'boleto', 'pix', 'cartao']


### RESPOSTA

Item (4A)

A codificação de variáveis categóricas é o processo de converter dados qualitativos (textos, categorias ou etiquetas) em formatos numéricos. A grande maioria dos algoritmos de machine learning (como Regressão Linear, SVM e Redes Neurais) é baseada em equações matemáticas e cálculos de distância. Como esses algoritmos não conseguem processar strings como "cartão" ou "pix" diretamente, precisamos transformar essas categorias em números para que o modelo possa realizar operações aritméticas e encontrar padrões nos dados.
Existem duas técnicas principais:
  * Label Encoding: Atribui um número inteiro para cada categoria (ex: 0, 1, 2). Útil para categorias com ordem lógica (ex: Perto, Médio, Longe).
  * One-Hot Encoding: Cria uma nova coluna binária ($0$ ou $1$) para cada categoria única. É ideal para categorias sem ordem hierárquica, como o tipo de pagamento.

```
# RESPOSTA Item (4B)
# Seu código abaixo:

import pandas as pd

# 1. Criando o dataset
data = {
    'cliente_id': [1, 2, 3, 4],
    'tipo_pagamento': ['cartao', 'boleto', 'pix', 'cartao']
}

df = pd.DataFrame(data)

# 2. Aplicando o One-Hot Encoding
# O parâmetro 'dtype=int' garante que o resultado seja 0 e 1 (em vez de True/False)
df_encoded = pd.get_dummies(df, columns=['tipo_pagamento'], dtype=int)

print(df_encoded)
```

## Questão 06 (1.0)

a) O que é normalização e qual seu objetivo no contexto de modelos preditivos?

b) No dataset abaixo, aplique normalização (Min-Max) na variável renda, transformando os valores para o intervalo [0,1].

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]


### RESPOSTA

Item (6A)

A normalização é uma técnica de escalonamento usada para colocar as variáveis numéricas dentro de um mesmo intervalo predefinido (geralmente de 0 a 1). Seu objetivo no contexto de modelos preditivos é padronizar a magnitude das variáveis para que features com valores absolutos maiores não dominem indevidamente os pesos do modelo (como na regressão) e também para facilitar e acelerar a convergência em algoritmos baseados em cálculo de distâncias.

```
# RESPOSTA Item (6B)
# Seu código abaixo:

import pandas as pd

# 1. Criando o dataset (Dicionário -> DataFrame)
data = {
    'cliente_id': [1, 2, 3, 4],
    'renda': [2000, 5000, 3000, 4000]
}

df = pd.DataFrame(data)

# 2. Definindo os valores mínimo e máximo da coluna 'renda'
renda_min = df['renda'].min()  # Retorna 2000
renda_max = df['renda'].max()  # Retorna 5000

# 3. Aplicando a fórmula: (x - min) / (max - min)
# Isso transformará todos os valores para o intervalo [0, 1]
df['renda_normalizada'] = (df['renda'] - renda_min) / (renda_max - renda_min)

# Exibindo o resultado final
print(df)

# Opcional: Salvando o resultado em um arquivo CSV
df.to_csv('dados_normalizados.csv', index=False)
```

## Questão 07 (1.0)

a) O que é padronização e qual a diferença em relação à normalização?

b) No dataset abaixo, aplique padronização (Z-score) na variável renda.

    'cliente_id': [1,2,3,4],
    'renda': [2000, 5000, 3000, 4000]

### RESPOSTA

Item (7A)

A padronização (ou Z-score normalization) é uma técnica de pré-processamento que reescala os dados para que eles tenham uma média igual a 0 e um desvio padrão igual a 1.
As principais diferenças são:
  * Escala de saída: normalização (Min-Max): Comprime os dados em um intervalo fixo, geralmente entre $[0, 1]$.Padronização (Z-score): Não possui um limite superior ou inferior fixo. Os valores resultantes indicam a quantos desvios padrões o dado original está da média.
  * Sensibilidade a Outliers: a normalização é muito sensível a outliers (valores atípicos), pois um único valor extremo pode "esmagar" todos os outros dados em um intervalo minúsculo.A padronização é muito mais robusta a outliers, pois não força os dados a um intervalo fixo.
  * Distribuição dos Dados: a padronização pressupõe (ou tenta aproximar) que os dados seguem uma distribuição Gaussiana (Normal). É a escolha preferida para algoritmos como Regressão Linear, Regressão Logística e SVM.

```
# RESPOSTA Item (7B)
# Seu código abaixo:

import pandas as pd
import numpy as np

# Criando o dataset
data = {
    'cliente_id': [1, 2, 3, 4],
    'renda': [2000, 5000, 3000, 4000]
}

df = pd.DataFrame(data)

# Calculando a média e desvio padrão
media = df['renda'].mean()
desvio_padrao = df['renda'].std(ddof=0) # ddof=0 para desvio padrão populacional

# Aplicando a fórmula do Z-score
df['renda_padronizada'] = (df['renda'] - media) / desvio_padrao

print(df)
```

## Questão 08 (1.0)

a) O que é seleção de atributos e qual seu objetivo em análise preditiva?

b) No dataset abaixo, identifique e remova as variáveis que não devem ser utilizadas no modelo e, em seguida, separe X e y.

    'cliente_id': [1,2,3,4],
    'idade': [25, 40, 35, 50],
    'renda_mensal': [2000, 5000, 3000, 4000],
    'renda_anual': [24000, 60000, 36000, 48000],
    'inadimplente': [0,1,0,1]

### RESPOSTA

Item (8A)

A seleção de atributos (feature selection) é o processo de selecionar um subconjunto de variáveis relevantes para serem usadas na construção de um modelo preditivo. Nem todos os dados disponíveis em um banco de dados são úteis para uma previsão; alguns podem ser irrelevantes, redundantes ou até prejudiciais.

Os principais objetivos são:

  * Reduzir o Overfitting: Ao remover dados irrelevantes (ruído), o modelo aprende padrões reais em vez de memorizar flutuações aleatórias.

  * Melhorar a Acurácia: Menos variáveis contraditórias ou redundantes permitem que o algoritmo identifique melhor as relações de causa e efeito.

  * Reduzir o Tempo de Treinamento: Menos dados significam menos cálculos, o que é crucial em grandes volumes de dados (Big Data).

  * Aumentar a Interpretabilidade: Modelos com menos variáveis são mais fáceis de explicar para partes interessadas.

```
# RESPOSTA Item (8B)
# Seu código abaixo:

import pandas as pd

# 1. Criando o dataset
data = {
    'cliente_id': [1, 2, 3, 4],
    'idade': [25, 40, 35, 50],
    'renda_mensal': [2000, 5000, 3000, 4000],
    'renda_anual': [24000, 60000, 36000, 48000],
    'inadimplente': [0, 1, 0, 1]
}

df = pd.DataFrame(data)

# 2. Removendo variáveis irrelevantes ou redundantes
# Removemos 'cliente_id' (ID) e 'renda_anual' (redundante/multicolinear)
df_limpo = df.drop(columns=['cliente_id', 'renda_anual'])

# 3. Separando X (Atributos/Features) e y (Alvo/Target)
X = df_limpo.drop(columns=['inadimplente']) # Tudo menos a resposta
y = df_limpo['inadimplente']               # Apenas a variável resposta

print("--- Matriz de Atributos (X) ---")
print(X)
print("\n--- Vetor Alvo (y) ---")
print(y)
```

## Questão 10 (1.0)

a) O que é informação mútua e como ela pode ser utilizada na seleção de atributos?

b) Crie o DataFrame abaixo, calcule a informação mútua entre as variáveis explicativas e a variável-alvo e identifique qual variável apresenta menor relevância. Em seguida, remova essa variável, separe X e y e exiba o resultado.

    "temperatura": [15,18,21,24,27,30,33,36,39],
    "marketing": [100,120,140,160,180,200,220,240,260],
    "renda_regiao": [2000,2100,2200,2300,2400,2500,2600,2700,2800],
    "vendas_sorvete": [180,250,330,420,520,630,750,880,1020]

### RESPOSTA

Item (10A)

Informação Mútua é um método para avaliar um par de variáveis a partir das probabilidades conjuntas e marginais de tal forma que quanto maior for a Informação Mútua maior será a dependência entre as variáveis.

Ela é usada em Engenharia de atributos para avaliar e remover variáveis redundantes ou irrelevantes.

```
# RESPOSTA Item (10B)
# Seu código abaixo:

import pandas as pd
from sklearn.feature_selection import mutual_info_regression
import numpy as np

df = pd.DataFrame({
    "temperatura": [15,18,21,24,27,30,33,36,39],
    "marketing": [100,120,140,160,180,200,220,240,260],
    "renda_regiao": [2000,2100,2200,2300,2400,2500,2600,2700,2800],
    "vendas_sorvete": [180,250,330,420,520,630,750,880,1020]
})

# Separar atributos e variável alvo.
X = df.drop(columns=["vendas_sorvete"])
y = df["vendas_sorvete"]

X.head(), y.head()

mi = mutual_info_regression(X, y, random_state=42)
mi_s = pd.Series(mi, index=X.columns).sort_values(ascending=False)

print(mi_s)

var_menos_rel= mi_s.idxmin()
print(f"Variável a ser removida: {var_menos_rel}")

mi_final= mi_s.drop(var_menos_rel)

print("\n--- Resultado Final ---")
print(mi_final)

print("\n--- Matriz de Atributos (X) ---")

print(X.drop(columns=[var_menos_rel]))

print("\n--- Vetor Alvo (y) ---")

print(y)
```
